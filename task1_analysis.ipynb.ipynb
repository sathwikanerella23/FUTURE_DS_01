```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Order Date": [
        "01-01-2024","02-01-2024","03-01-2024","04-01-2024","05-01-2024",
        "06-01-2024","07-01-2024","08-01-2024","09-01-2024","10-01-2024"
    ],

    "Product": [
        "Laptop","Phone","Shoes","Watch","Laptop",
        "Shirt","Headphones","Bag","Phone","Shoes"
    ],

    "Category": [
        "Electronics","Electronics","Fashion","Accessories","Electronics",
        "Fashion","Electronics","Accessories","Electronics","Fashion"
    ],

    "Region": [
        "South","North","East","West","South",
        "North","East","West","South","North"
    ],

    "Quantity": [
        2,3,5,4,1,
        6,3,2,2,4
    ],

    "Sales": [
        80000,60000,10000,8000,40000,
        12000,9000,4000,40000,8000
    ]
}
```

```python
df = pd.DataFrame(data)

df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

print(df)
```

```python
total_sales = df['Sales'].sum()
total_quantity = df['Quantity'].sum()

print("="*40)
print("        KPI ANALYSIS")
print("="*40)

print("Total Sales :", total_sales)
print("Total Quantity :", total_quantity)
```

```python
top_products = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

print("="*40)
print("     TOP SELLING PRODUCTS")
print("="*40)

print(top_products)
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print("="*40)
print("     CATEGORY PERFORMANCE")
print("="*40)

print(category_sales)
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

print("="*40)
print("      REGION PERFORMANCE")
print("="*40)

print(region_sales)
```

```python
sales_trend = df.groupby('Order Date')['Sales'].sum()

plt.figure(figsize=(10,5))

sales_trend.plot(marker='o')

plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

top_products.plot(kind='bar')

plt.title("Top Selling Products")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(7,7))

category_sales.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Category Contribution")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(kind='bar')

plt.title("Region Performance")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

## Insights

- Electronics category generates highest sales

- South region contributes more revenue

- Laptop and Phone dominate total sales

- Sales show variation over time

- Accessories category has lower sales


## Recommendations

- Focus more on high-performing categories

- Improve strategies in low-performing regions

- Promote top-selling products

- Increase stock for high-demand items

- Analyze sales trends regularly
